```python
def get_phi_embedding(text):
    messages = [
        {
            "role": "user",
            "content": text
        }
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=False,
        tokenize=True,
        return_dict=True,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model(
            **inputs,
            output_hidden_states=True,
            use_cache=False
        )

    hidden = outputs.hidden_states[-1]

    last_token = hidden[:, -1, :]

    return last_token.float().cpu()


x = get_phi_embedding(
    "Explain why quantum computers are different from classical computers."
)

print(x.shape)
print(model.config.hidden_size)
```

```python
import hashlib
import os
import torch

PHI_CACHE = os.path.join(
    BASE,
    "embeddings/phi"
)

def text_hash(text):
    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()[:32]


def get_cached_phi_embedding(text):

    h = text_hash(text)

    path = os.path.join(
        PHI_CACHE,
        h + ".pt"
    )

    if os.path.exists(path):
        return torch.load(path, map_location="cpu")

    embedding = get_phi_embedding(text)

    torch.save(
        embedding,
        path
    )

    return embedding

e = get_cached_phi_embedding(
    "Explain quantum entanglement."
)

print(e.shape)
```

```python
import numpy as np
import torch

N_QUBITS = 8
phi_dim = model.config.hidden_size

projection = torch.nn.Linear(
    phi_dim,
    N_QUBITS,
    bias=True
)

torch.manual_seed(SEED)

projection.eval()
```

```python
def phi_to_angles(phi_embedding):

    with torch.no_grad():
        x = projection(phi_embedding)

    x = torch.tanh(x)

    angles = x * np.pi

    return angles.squeeze(0).numpy()

angles = phi_to_angles(e)

print(angles)
print("Number of angles:", len(angles))
```

```python
from qiskit import QuantumCircuit

def build_q8_circuit(angles, layers=3):

    qc = QuantumCircuit(8)

    # Data encoding
    for i in range(8):
        qc.ry(float(angles[i]), i)
        qc.rz(float(angles[i]), i)

    # Variational / mixing layers
    for layer in range(layers):

        for i in range(8):
            qc.ry(
                0.25 * (layer + 1),
                i
            )

        # Ring entanglement
        for i in range(8):
            qc.cx(
                i,
                (i + 1) % 8
            )

    return qc

qc = build_q8_circuit(
    angles,
    CONFIG["circuit_layers"]
)

print(qc)
```

```python
from qiskit.quantum_info import Statevector, Pauli

def quantum_embedding(angles):

    qc = build_q8_circuit(
        angles,
        CONFIG["circuit_layers"]
    )

    state = Statevector.from_instruction(qc)

    features = []

    for basis in ["X", "Y", "Z"]:

        for i in range(8):

            label = ["I"] * 8
            label[7 - i] = basis

            pauli = Pauli("".join(label))

            value = np.real(
                state.expectation_value(pauli)
            )

            features.append(value)

    return np.array(features, dtype=np.float32)

q = quantum_embedding(angles)

print("Quantum embedding shape:", q.shape)
print(q)
```

```python
QUANTUM_CACHE = os.path.join(
    BASE,
    "embeddings/quantum"
)

def get_cached_quantum_embedding(text):

    h = text_hash(text)

    path = os.path.join(
        QUANTUM_CACHE,
        h + ".npy"
    )

    if os.path.exists(path):
        return np.load(path)

    phi = get_cached_phi_embedding(text)

    angles = phi_to_angles(phi)

    q = quantum_embedding(angles)

    np.save(path, q)

    return q
```

```python
from tqdm.auto import tqdm

records = []

for item in tqdm(dataset):

    text = item["text"]
    label = item["label"]

    phi = get_cached_phi_embedding(text)
    quantum = get_cached_quantum_embedding(text)

    records.append({
        "text": text,
        "label": label,
        "phi": phi.squeeze(0).numpy(),
        "quantum": quantum
    })

print("Processed:", len(records))
```

```python
classical_projection = torch.nn.Linear(
    phi_dim,
    24,
    bias=True
)

torch.manual_seed(SEED)

classical_projection.eval()

def classical_embedding(phi):

    with torch.no_grad():
        return classical_projection(phi).squeeze(0).numpy()

random_projection = torch.randn(
    phi_dim,
    24
) / np.sqrt(phi_dim)

def random_embedding(phi):

    x = phi.squeeze(0).numpy()

    return x @ random_projection.numpy()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

labels = encoder.fit_transform(
    [r["label"] for r in records]
)

X_quantum = np.stack([
    r["quantum"]
    for r in records
])

X_phi = torch.cat([
    torch.tensor(r["phi"]).unsqueeze(0)
    for r in records
])

X_classical = np.stack([
    classical_embedding(
        torch.tensor(r["phi"]).unsqueeze(0)
    )
    for r in records
])

X_random = np.stack([
    random_embedding(
        torch.tensor(r["phi"]).unsqueeze(0)
    )
    for r in records
])

print(X_quantum.shape)
print(X_classical.shape)
print(X_random.shape)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

def evaluate_embedding(X, y):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.3,
        random_state=SEED,
        stratify=y
    )

    clf = LogisticRegression(
        max_iter=2000,
        random_state=SEED
    )

    clf.fit(X_train, y_train)

    predictions = clf.predict(X_test)

    return accuracy_score(
        y_test,
        predictions
    )



results = {
    "classical": evaluate_embedding(
        X_classical,
        labels
    ),

    "random": evaluate_embedding(
        X_random,
        labels
    ),

    "quantum": evaluate_embedding(
        X_quantum,
        labels
    )
}

results
```
